### Data Cleaning

```python
new = data.drop(columns=['id', 'name', 'host_id', 'host_name', 'last_review'], axis = 1)
new['reviews_per_month'] = new['reviews_per_month'].fillna(0)
```

```python
price_cap  = new['price'].quantile(0.99)
nights_cap = new['minimum_nights'].quantile(0.99)

new['price'] = new['price'].clip(upper=price_cap)
new['minimum_nights'] = new['minimum_nights'].clip(upper=nights_cap)
```

```python
sns.barplot(x='minimum_nights', y='price', data=new)
```

```python
X = new.drop(columns=['room_type'])
y = new['room_type']
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42, stratify=y)
```

### Preprocessing (ColumnTransformer and Pipeline)

```python
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder, PowerTransformer
from sklearn.compose import ColumnTransformer
```

```python
numerical_cols = ["latitude","longitude","price", "minimum_nights", "number_of_reviews",
                 "reviews_per_month", "calculated_host_listings_count",
                 "availability_365"]

categorical_cols = ['neighbourhood_group', 'neighbourhood']

#1. Pipeline -> for Numeric Columns
numeric_pipeline = Pipeline(steps=[
    ('impute', SimpleImputer(strategy="median")),
    ('skewness', PowerTransformer(method='yeo-johnson', standardize=True)),
    ('scale', StandardScaler())
])


#2. Pipeline -> for Categorical Columns
categorical_pipeline = Pipeline(steps=[
    ('impute', SimpleImputer(strategy="most_frequent")),
    ('encode', OneHotEncoder(handle_unknown='ignore'))
])

preprocessor = ColumnTransformer(transformers=[
    ("numerical", numeric_pipeline, numerical_cols),
    ("categorical", categorical_pipeline, categorical_cols )
])
```

### Multiple machine learning models

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.model_selection import cross_val_score
```

```python
models = {
    "Logistic Regression" : LogisticRegression(class_weight="balanced",random_state=42),
    "Decision Tree"       : DecisionTreeClassifier(class_weight="balanced",random_state=42),
    "Random Forest"       : RandomForestClassifier(class_weight="balanced", random_state=42),
    "Gradient Boosting"   : GradientBoostingClassifier(random_state=42),
}
```

```python
for name, model in models.items():
    pipe = Pipeline(steps=[("preprocessor", preprocessor), ("classifier", model)])

    accuracy = cross_val_score(pipe, X_train, y_train, cv=3, scoring='accuracy') #accuracy alone can be misleading in imbalanced datasets, so we will use macro f1 score

    macrof1 = cross_val_score(pipe, X_train, y_train, cv=3, scoring='f1_macro')

    print(f'{name} -> Accuracy: {accuracy.mean():.2f}, F1: {macrof1.mean():.2f}')
```

### Hyperparameter Tuning

```python
from sklearn.model_selection import RandomizedSearchCV

best_pipeline = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("classifier",   RandomForestClassifier(class_weight='balanced'))
])

param_distribution = {
    "classifier__n_estimators" : [100,200,150,300],
    "classifier__max_depth": [8, 12, 15, 20, None],
    "classifier__min_samples_split": [2, 5, 10],
}

search = RandomizedSearchCV(
    estimator=best_pipeline,
    param_distributions=param_distribution,
    n_iter=10,
    cv=3,
    scoring="f1_macro",
    random_state=42
    )

search.fit(X_train, y_train)
```

### Final evaluation 

```python
print("Best Parameters: ", search.best_params_)
print("Best CV Macro-F1: ", search.best_score_)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix,f1_score
best_pipeline = search.best_estimator_

y_pred = best_pipeline.predict(X_test)

print(f"Accuracy Score: {accuracy_score(y_test, y_pred)}")
print(f"F1 Score: {f1_score(y_test, y_pred, average="macro")}")
```

```python
sns.heatmap(confusion_matrix(y_test,y_pred), annot=True, fmt=".2f",
            xticklabels=best_pipeline.classes_, yticklabels=best_pipeline.classes_)
```
